## `Visualization`

# Type 1 : Age Distribution

```python
plt.figure(figsize=(8, 5))

sns.kdeplot(df['Age'], fill=True)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Density')

plt.show()
```

# Type 2 : Age Distribution

```python
plt.figure(figsize=(10, 5))

sns.countplot(x=df['Age'])

plt.title('Age Frequency')
plt.xlabel('Age')
plt.ylabel('Count')

plt.show()
```

# Gender Distribution

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(x='Gender', data=df)

total = len(df)

for p in ax.patches:
    percentage = 100 * p.get_height() / total
    ax.annotate(
        f'{percentage:.1f}%',
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha='center',
        va='bottom'
    )

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.show()
```

# Workout Type Distribution

```python
plt.figure(figsize=(8, 5))

ax = sns.countplot(x='Workout_Type', data=df)

total = len(df)

for p in ax.patches:
    percentage = 100 * p.get_height() / total
    ax.annotate(
        f'{percentage:.1f}%',
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha='center',
        va='bottom'
    )

plt.title('Workout Type Distribution')
plt.xlabel('Workout Type')
plt.ylabel('Count')

plt.show()
```
